```python
%%writefile /kaggle/working/train_qwen_ddp.py
import os

os.environ["PYTORCH_CUDA_ALLOC_CONF"] = "expandable_segments:True"

import torch

from datasets import load_from_disk

from transformers import (
    AutoTokenizer,
    AutoModelForCausalLM,
    BitsAndBytesConfig,
    TrainingArguments,
    Trainer,
)

from peft import (
    LoraConfig,
    get_peft_model,
)


# ============================================================
# PATHS
# ============================================================

MODEL_PATH = (
    "/kaggle/working/Qwen2.5-7B-Instruct"
)

TRAIN_PATH = (
    "/kaggle/working/qwen_train"
)

OUTPUT_DIR = (
    "/kaggle/working/llm-judge-stage1"
)


# ============================================================
# SETTINGS
# ============================================================

MAX_LENGTH = 768

BATCH_SIZE = 1

GRADIENT_ACCUMULATION_STEPS = 8

LEARNING_RATE = 1e-4

NUM_EPOCHS = 1

WARMUP_STEPS = 300


# ============================================================
# SYSTEM PROMPT
# ============================================================

SYSTEM_PROMPT = """You are an impartial AI response judge.

Given a question and two candidate answers, determine which answer is better.

Evaluate the answers based on:
- correctness
- relevance
- completeness
- reasoning quality
- clarity

Do not consider the identity or reputation of the model that produced an answer.

Return only JSON in this format:

{"winner":"A"}

The winner must be exactly one of:
A
B
tie
"""


# ============================================================
# COLLATOR
# ============================================================

class JudgeDataCollator:

    def __init__(self, tokenizer):
        self.tokenizer = tokenizer

    def __call__(self, features):

        input_ids = [
            x["input_ids"]
            for x in features
        ]

        attention_masks = [
            x["attention_mask"]
            for x in features
        ]

        labels = [
            x["labels"]
            for x in features
        ]

        batch = self.tokenizer.pad(
            {
                "input_ids": input_ids,
                "attention_mask": attention_masks,
            },
            padding=True,
            pad_to_multiple_of=8,
            return_tensors="pt",
        )

        max_len = batch["input_ids"].shape[1]

        padded_labels = []

        for label in labels:

            padded_labels.append(
                label + [-100] * (
                    max_len - len(label)
                )
            )

        batch["labels"] = torch.tensor(
            padded_labels,
            dtype=torch.long,
        )

        return batch


# ============================================================
# TOKENIZATION
# ============================================================

def tokenize_example(example, tokenizer):

    messages = example["messages"]

    prompt_messages = messages[:2]

    full_messages = messages

    prompt_text = tokenizer.apply_chat_template(
        prompt_messages,
        tokenize=False,
        add_generation_prompt=True,
    )

    full_text = tokenizer.apply_chat_template(
        full_messages,
        tokenize=False,
        add_generation_prompt=False,
    )

    # IMPORTANT:
    # Limit prompt tokenization too.
    prompt_ids = tokenizer(
        prompt_text,
        add_special_tokens=False,
        truncation=True,
        max_length=MAX_LENGTH,
    )["input_ids"]

    full = tokenizer(
        full_text,
        add_special_tokens=False,
        truncation=True,
        max_length=MAX_LENGTH,
    )

    input_ids = full["input_ids"]

    attention_mask = full["attention_mask"]

    prompt_length = min(
        len(prompt_ids),
        len(input_ids),
    )

    labels = input_ids.copy()

    for i in range(prompt_length):
        labels[i] = -100

    # Safety check
    if all(x == -100 for x in labels):
        raise ValueError(
            "Example contains no trainable assistant tokens."
        )

    return {
        "input_ids": input_ids,
        "attention_mask": attention_mask,
        "labels": labels,
    }


# ============================================================
# MAIN
# ============================================================

def main():

    local_rank = int(
        os.environ.get("LOCAL_RANK", "0")
    )

    torch.cuda.set_device(local_rank)

    print("=" * 60)
    print("DISTRIBUTED GPU")
    print("=" * 60)

    print("LOCAL_RANK:", local_rank)
    print(
        "GPU:",
        torch.cuda.get_device_name(local_rank),
    )

    print(
        "GPU COUNT:",
        torch.cuda.device_count(),
    )

    # --------------------------------------------------------
    # TOKENIZER
    # --------------------------------------------------------

    print("\n" + "=" * 60)
    print("LOADING TOKENIZER")
    print("=" * 60)

    tokenizer = AutoTokenizer.from_pretrained(
        MODEL_PATH,
        local_files_only=True,
        trust_remote_code=True,
        use_fast=True,
    )

    if tokenizer.pad_token is None:
        tokenizer.pad_token = tokenizer.eos_token

    tokenizer.padding_side = "right"

    # --------------------------------------------------------
    # DATASET
    # --------------------------------------------------------

    print("\n" + "=" * 60)
    print("LOADING DATASET")
    print("=" * 60)

    dataset = load_from_disk(TRAIN_PATH)

    print(
        "Training examples:",
        len(dataset),
    )

    # --------------------------------------------------------
    # TOKENIZATION
    # --------------------------------------------------------

    print("\n" + "=" * 60)
    print("TOKENIZING")
    print("=" * 60)

    tokenized = dataset.map(
        lambda x: tokenize_example(
            x,
            tokenizer,
        ),
        remove_columns=dataset.column_names,
        desc=f"Tokenizing rank {local_rank}",
    )

    print(
        "Tokenization complete:",
        len(tokenized),
    )

    # --------------------------------------------------------
    # 4-BIT
    # --------------------------------------------------------

    print("\n" + "=" * 60)
    print("CONFIGURING 4-BIT")
    print("=" * 60)

    bnb_config = BitsAndBytesConfig(
        load_in_4bit=True,
        bnb_4bit_quant_type="nf4",
        bnb_4bit_use_double_quant=True,
        bnb_4bit_compute_dtype=torch.float16,
    )

    # --------------------------------------------------------
    # LOAD MODEL
    # --------------------------------------------------------

    print("\n" + "=" * 60)
    print("LOADING QWEN")
    print("=" * 60)

    model = AutoModelForCausalLM.from_pretrained(
        MODEL_PATH,
        quantization_config=bnb_config,
        device_map={
            "": local_rank
        },
        dtype=torch.float16,
        local_files_only=True,
        low_cpu_mem_usage=True,
        trust_remote_code=True,
    )

    model.config.use_cache = False

    print(
        "Qwen loaded on GPU:",
        local_rank,
    )

    print(
        "GPU memory:",
        round(
            torch.cuda.memory_allocated(local_rank)
            / (1024 ** 3),
            2,
        ),
        "GB",
    )

    # --------------------------------------------------------
    # LoRA
    # --------------------------------------------------------

    print("\n" + "=" * 60)
    print("CONFIGURING LoRA")
    print("=" * 60)

    lora_config = LoraConfig(
        r=8,
        lora_alpha=16,
        lora_dropout=0.05,
        bias="none",
        task_type="CAUSAL_LM",
        target_modules=[
            "q_proj",
            "k_proj",
            "v_proj",
            "o_proj",
        ],
    )

    model = get_peft_model(
        model,
        lora_config,
    )

    model.enable_input_require_grads()

    model.print_trainable_parameters()

    # --------------------------------------------------------
    # TRAINING
    # --------------------------------------------------------

    print("\n" + "=" * 60)
    print("TRAINING CONFIGURATION")
    print("=" * 60)

    training_args = TrainingArguments(

        output_dir=OUTPUT_DIR,

        num_train_epochs=NUM_EPOCHS,

        per_device_train_batch_size=BATCH_SIZE,

        gradient_accumulation_steps=(
            GRADIENT_ACCUMULATION_STEPS
        ),

        learning_rate=LEARNING_RATE,

        weight_decay=0.01,

        fp16=True,

        gradient_checkpointing=True,

        optim="paged_adamw_8bit",

        lr_scheduler_type="cosine",

        warmup_steps=WARMUP_STEPS,

        logging_steps=10,

        save_strategy="steps",

        save_steps=2000,

        save_total_limit=1,

        report_to="none",

        remove_unused_columns=False,

        max_grad_norm=0.3,

        ddp_find_unused_parameters=False,
    )

    trainer = Trainer(
        model=model,
        args=training_args,
        train_dataset=tokenized,
        data_collator=JudgeDataCollator(
            tokenizer,
        ),
    )

    # --------------------------------------------------------
    # TRAIN
    # --------------------------------------------------------

    print("\n" + "=" * 60)
    print("STARTING 2xT4 QLORA TRAINING")
    print("=" * 60)

    trainer.train()

    # --------------------------------------------------------
    # SAVE
    # --------------------------------------------------------

    if local_rank == 0:

        print("\n" + "=" * 60)
        print("SAVING LoRA ADAPTER")
        print("=" * 60)

        trainer.save_model(
            OUTPUT_DIR
        )

        tokenizer.save_pretrained(
            OUTPUT_DIR
        )

        print(
            "Adapter saved to:",
            OUTPUT_DIR,
        )


if __name__ == "__main__":
    main()
```

```python
%%writefile /kaggle/working/train_qwen_ddp.py

import os

os.environ["PYTORCH_CUDA_ALLOC_CONF"] = "expandable_segments:True"

import json
import time
import shutil
import re

import torch

from datasets import load_from_disk

from transformers import (
    AutoTokenizer,
    AutoModelForCausalLM,
    BitsAndBytesConfig,
    TrainingArguments,
    Trainer,
    TrainerCallback,
)

from peft import (
    LoraConfig,
    get_peft_model,
)

from kaggle_secrets import UserSecretsClient

from google.oauth2.credentials import Credentials
from googleapiclient.discovery import build
from googleapiclient.http import MediaFileUpload, MediaIoBaseDownload

import io


# ============================================================
# PATHS
# ============================================================

MODEL_PATH = "/kaggle/working/Qwen2.5-7B-Instruct"

TRAIN_PATH = "/kaggle/working/qwen_train"

OUTPUT_DIR = "/kaggle/working/llm-judge-stage1"


# ============================================================
# GOOGLE DRIVE
# ============================================================

STAGE1_FOLDER_ID = "1rtwK-QEMRsQ8oZhNt07HBGbZHYincCPK"


# ============================================================
# TRAINING SETTINGS
# ============================================================

MAX_LENGTH = 768

BATCH_SIZE = 1

GRADIENT_ACCUMULATION_STEPS = 8

LEARNING_RATE = 1e-4

NUM_EPOCHS = 1

WARMUP_STEPS = 300

SAVE_STEPS = 250


# ============================================================
# SYSTEM PROMPT
# ============================================================

SYSTEM_PROMPT = """You are an impartial AI response judge.

Given a question and two candidate answers, determine which answer is better.

Evaluate the answers based on:
- correctness
- relevance
- completeness
- reasoning quality
- clarity

Do not consider the identity or reputation of the model that produced an answer.

Return only JSON in this format:

{"winner":"A"}

The winner must be exactly one of:
A
B
tie
"""


# ============================================================
# DRIVE AUTHENTICATION
# ============================================================

def get_drive_service():

    secrets = UserSecretsClient()

    client_id = secrets.get_secret("CLIENT_ID")
    client_secret = secrets.get_secret("CLIENT_SECRET")
    refresh_token = secrets.get_secret("REFRESH_TOKEN")

    creds = Credentials(
        token=None,
        refresh_token=refresh_token,
        token_uri="https://oauth2.googleapis.com/token",
        client_id=client_id,
        client_secret=client_secret,
        scopes=["https://www.googleapis.com/auth/drive"],
    )

    drive = build(
        "drive",
        "v3",
        credentials=creds,
    )

    return drive


# ============================================================
# DRIVE HELPERS
# ============================================================

def create_drive_folder(drive, name, parent_id):

    metadata = {
        "name": name,
        "mimeType": "application/vnd.google-apps.folder",
        "parents": [parent_id],
    }

    result = drive.files().create(
        body=metadata,
        fields="id,name",
    ).execute()

    return result["id"]


def upload_file(drive, local_path, parent_id, drive_name=None):

    if not os.path.exists(local_path):
        raise FileNotFoundError(local_path)

    if drive_name is None:
        drive_name = os.path.basename(local_path)

    metadata = {
        "name": drive_name,
        "parents": [parent_id],
    }

    media = MediaFileUpload(
        local_path,
        resumable=True,
    )

    request = drive.files().create(
        body=metadata,
        media_body=media,
        fields="id,name,size",
    )

    response = None

    while response is None:

        try:

            status, response = request.next_chunk()

            if status:
                percent = int(
                    status.progress() * 100
                )

                print(
                    f"Uploading {drive_name}: "
                    f"{percent}%",
                    flush=True,
                )

        except Exception as e:

            print(
                f"Upload interrupted: {e}"
            )

            time.sleep(5)

    return response


def upload_checkpoint_to_drive(
    checkpoint_path,
    global_step,
):

    drive = get_drive_service()

    checkpoint_name = f"checkpoint-{global_step}"

    print()
    print("=" * 70)
    print(f"DRIVE UPLOAD: {checkpoint_name}")
    print("=" * 70)

    # --------------------------------------------------------
    # Create checkpoint folder
    # --------------------------------------------------------

    checkpoint_folder_id = create_drive_folder(
        drive,
        checkpoint_name,
        STAGE1_FOLDER_ID,
    )

    uploaded_files = []

    # --------------------------------------------------------
    # Upload checkpoint files
    # --------------------------------------------------------

    for root, _, files in os.walk(checkpoint_path):

        for filename in files:

            local_file = os.path.join(
                root,
                filename,
            )

            relative_path = os.path.relpath(
                local_file,
                checkpoint_path,
            )

            # Checkpoints are normally flat.
            # Replace separators just in case.
            drive_name = relative_path.replace(
                os.sep,
                "__",
            )

            upload_file(
                drive,
                local_file,
                checkpoint_folder_id,
                drive_name,
            )

            uploaded_files.append(
                relative_path
            )

    # --------------------------------------------------------
    # Manifest
    # --------------------------------------------------------

    manifest = {
        "checkpoint": checkpoint_name,
        "global_step": global_step,
        "files": uploaded_files,
        "timestamp": time.time(),
    }

    manifest_path = "/tmp/checkpoint_manifest.json"

    with open(
        manifest_path,
        "w",
    ) as f:

        json.dump(
            manifest,
            f,
            indent=2,
        )

    upload_file(
        drive,
        manifest_path,
        checkpoint_folder_id,
        "manifest.json",
    )

    # --------------------------------------------------------
    # COMPLETE MARK
```

```python
%%writefile /kaggle/working/train_qwen_ddp.py

import os

os.environ["PYTORCH_CUDA_ALLOC_CONF"] = "expandable_segments:True"

import io
import json
import re
import shutil
import time

import torch

from datasets import load_from_disk

from transformers import (
    AutoTokenizer,
    AutoModelForCausalLM,
    BitsAndBytesConfig,
    TrainingArguments,
    Trainer,
    TrainerCallback,
)

from peft import (
    LoraConfig,
    get_peft_model,
)

from kaggle_secrets import UserSecretsClient

from google.oauth2.credentials import Credentials
from googleapiclient.discovery import build
from googleapiclient.http import (
    MediaFileUpload,
    MediaIoBaseDownload,
)


# ============================================================
# PATHS
# ============================================================

MODEL_PATH = "/kaggle/working/Qwen2.5-7B-Instruct"

TRAIN_PATH = "/kaggle/working/qwen_train"

OUTPUT_DIR = "/kaggle/working/llm-judge-stage1"


# ============================================================
# GOOGLE DRIVE
# ============================================================

STAGE1_FOLDER_ID = (
    "1rtwK-QEMRsQ8oZhNt07HBGbZHYincCPK"
)


# ============================================================
# TRAINING SETTINGS
# ============================================================

MAX_LENGTH = 768

BATCH_SIZE = 1

GRADIENT_ACCUMULATION_STEPS = 8

LEARNING_RATE = 1e-4

NUM_EPOCHS = 1

WARMUP_STEPS = 300

SAVE_STEPS = 250


# ============================================================
# SYSTEM PROMPT
# ============================================================

SYSTEM_PROMPT = """You are an impartial AI response judge.

Given a question and two candidate answers, determine which answer is better.

Evaluate the answers based on:
- correctness
- relevance
- completeness
- reasoning quality
- clarity

Do not consider the identity or reputation of the model that produced an answer.

Return only JSON in this format:

{"winner":"A"}

The winner must be exactly one of:
A
B
tie
"""


# ============================================================
# GOOGLE DRIVE AUTHENTICATION
# ============================================================

def get_drive_service():

    secrets = UserSecretsClient()

    client_id = secrets.get_secret(
        "CLIENT_ID"
    )

    client_secret = secrets.get_secret(
        "CLIENT_SECRET"
    )

    refresh_token = secrets.get_secret(
        "REFRESH_TOKEN"
    )

    creds = Credentials(
        token=None,
        refresh_token=refresh_token,
        token_uri="https://oauth2.googleapis.com/token",
        client_id=client_id,
        client_secret=client_secret,
        scopes=[
            "https://www.googleapis.com/auth/drive"
        ],
    )

    drive = build(
        "drive",
        "v3",
        credentials=creds,
    )

    return drive


# ============================================================
# CREATE DRIVE FOLDER
# ============================================================

def create_drive_folder(
    drive,
    name,
    parent_id,
):

    metadata = {
        "name": name,
        "mimeType": (
            "application/vnd.google-apps.folder"
        ),
        "parents": [parent_id],
    }

    result = drive.files().create(
        body=metadata,
        fields="id,name",
    ).execute()

    return result["id"]


# ============================================================
# UPLOAD FILE
# ============================================================

def upload_file(
    drive,
    local_path,
    parent_id,
    drive_name=None,
):

    if not os.path.exists(local_path):
        raise FileNotFoundError(
            local_path
        )

    if drive_name is None:
        drive_name = os.path.basename(
            local_path
        )

    metadata = {
        "name": drive_name,
        "parents": [parent_id],
    }

    media = MediaFileUpload(
        local_path,
        resumable=True,
    )

    request = drive.files().create(
        body=metadata,
        media_body=media,
        fields="id,name,size",
    )

    response = None

    while response is None:

        try:

            status, response = (
                request.next_chunk()
            )

            if status:

                percent = int(
                    status.progress() * 100
                )

                print(
                    f"Uploading {drive_name}: "
                    f"{percent}%",
                    flush=True,
                )

        except Exception as e:

            print(
                f"Upload interrupted: {e}"
            )

            time.sleep(5)

    print(
        f"Uploaded: {drive_name}"
    )

    return response


# ============================================================
# UPLOAD CHECKPOINT
# ============================================================

def upload_checkpoint_to_drive(
    checkpoint_path,
    global_step,
):

    drive = get_drive_service()

    checkpoint_name = (
        f"checkpoint-{global_step}"
    )

    print()
    print("=" * 70)
    print(
        f"DRIVE UPLOAD: {checkpoint_name}"
    )
    print("=" * 70)

    checkpoint_folder_id = (
        create_drive_folder(
            drive,
            checkpoint_name,
            STAGE1_FOLDER_ID,
        )
    )

    uploaded_files = []

    # --------------------------------------------------------
    # Upload every checkpoint file
    # --------------------------------------------------------

    for root, _, files in os.walk(
        checkpoint_path
    ):

        for filename in files:

            local_file = os.path.join(
                root,
                filename,
            )

            relative_path = os.path.relpath(
                local_file,
                checkpoint_path,
            )

            drive_name = relative_path.replace(
                os.sep,
                "__",
            )

            upload_file(
                drive,
                local_file,
                checkpoint_folder_id,
                drive_name,
            )

            uploaded_files.append(
                relative_path
            )

    # --------------------------------------------------------
    # Manifest
    # --------------------------------------------------------

    manifest = {
        "checkpoint": checkpoint_name,
        "global_step": global_step,
        "files": uploaded_files,
        "timestamp": time.time(),
    }

    manifest_path = (
        "/tmp/checkpoint_manifest.json"
    )

    with open(
        manifest_path,
        "w",
    ) as f:

        json.dump(
            manifest,
            f,
            indent=2,
        )

    upload_file(
        drive,
        manifest_path,
        checkpoint_folder_id,
        "manifest.json",
    )

    # --------------------------------------------------------
    # COMPLETE MARKER
    # --------------------------------------------------------

    marker_path = "/tmp/_COMPLETE"

    with open(
        marker_path,
        "w",
    ) as f:

        f.write(
            f"checkpoint-{global_step}\n"
        )

    # _COMPLETE is uploaded LAST.
    upload_file(
        drive,
        marker_path,
        checkpoint_folder_id,
        "_COMPLETE",
    )

    print()
    print(
        f"✅ DRIVE CHECKPOINT COMPLETE: "
        f"{checkpoint_name}"
    )
    print()


# ============================================================
# FIND LATEST COMPLETE CHECKPOINT
# ============================================================

def find_latest_complete_checkpoint():

    drive = get_drive_service()

    result = drive.files().list(
        q=(
            f"'{STAGE1_FOLDER_ID}' in parents "
            f"and mimeType = "
            f"'application/vnd.google-apps.folder' "
            f"and trashed = false"
        ),
        fields="files(id,name)",
        pageSize=1000,
    ).execute()

    folders = result.get(
        "files",
        []
    )

    candidates = []

    for folder in folders:

        match = re.fullmatch(
            r"checkpoint-(\d+)",
            folder["name"],
        )

        if not match:
            continue

        step = int(
            match.group(1)
        )

        marker_result = (
            drive.files().list(
                q=(
                    f"'{folder['id']}' in parents "
                    f"and name = '_COMPLETE' "
                    f"and trashed = false"
                ),
                fields="files(id,name)",
                pageSize=10,
            ).execute()
        )

        if marker_result.get(
            "files",
            []
        ):

            candidates.append(
                (
                    step,
                    folder["id"],
                    folder["name"],
                )
            )

    if not candidates:

        print(
            "No complete Drive checkpoint found."
        )

        return None

    candidates.sort(
        key=lambda x: x[0],
        reverse=True,
    )

    latest = candidates[0]

    print(
        f"Latest complete checkpoint: "
        f"{latest[2]}"
    )

    return latest


# ============================================================
# DOWNLOAD LATEST CHECKPOINT
# ============================================================

def download_latest_checkpoint():

    latest = (
        find_latest_complete_checkpoint()
    )

    if latest is None:
        return None

    step, folder_id, folder_name = latest

    local_path = os.path.join(
        OUTPUT_DIR,
        folder_name,
    )

    if os.path.exists(local_path):

        shutil.rmtree(
            local_path
        )

    os.makedirs(
        local_path,
        exist_ok=True,
    )

    drive = get_drive_service()

    result = drive.files().list(
        q=(
            f"'{folder_id}' in parents "
            f"and trashed = false"
        ),
        fields="files(id,name,size)",
        pageSize=1000,
    ).execute()

    files = result.get(
        "files",
        []
    )

    print()
    print(
        f"Downloading {folder_name}"
    )

    for file in files:

        name = file["name"]

        if name in {
            "_COMPLETE",
            "manifest.json",
        }:

            continue

        local_file = os.path.join(
            local_path,
            name,
        )

        request = drive.files().get_media(
            fileId=file["id"]
        )

        with open(
            local_file,
            "wb",
        ) as f:

            downloader = (
                MediaIoBaseDownload(
                    f,
                    request,
                )
            )

            done = False

            while not done:

                status, done = (
                    downloader.next_chunk()
                )

                if status:

                    percent = int(
                        status.progress() * 100
                    )

                    print(
                        f"Downloading {name}: "
                        f"{percent}%",
                        flush=True,
                    )

    print()
    print(
        f"✅ Checkpoint downloaded: "
        f"{local_path}"
    )

    return local_path


# ============================================================
# DATA COLLATOR
# ============================================================

class JudgeDataCollator:

    def __init__(
        self,
        tokenizer,
    ):

        self.tokenizer = tokenizer

    def __call__(
        self,
        features,
    ):

        input_ids = [
            x["input_ids"]
            for x in features
        ]

        attention_masks = [
            x["attention_mask"]
            for x in features
        ]

        labels = [
            x["labels"]
            for x in features
        ]

        batch = self.tokenizer.pad(
            {
                "input_ids": input_ids,
                "attention_mask": attention_masks,
            },
            padding=True,
            pad_to_multiple_of=8,
            return_tensors="pt",
        )

        max_len = batch[
            "input_ids"
        ].shape[1]

        padded_labels = []

        for label in labels:

            padded_labels.append(
                label + [-100] * (
                    max_len - len(label)
                )
            )

        batch["labels"] = torch.tensor(
            padded_labels,
            dtype=torch.long,
        )

        return batch


# ============================================================
# CORRECT TOKENIZATION
# ============================================================

def tokenize_example(
    example,
    tokenizer,
):

    messages = example["messages"]

    # --------------------------------------------------------
    # We assume the final message is the judge answer.
    # Everything before it is the prompt.
    # --------------------------------------------------------

    if len(messages) < 2:

        raise ValueError(
            "Example must contain at least "
            "one prompt and one response."
        )

    prompt_messages = messages[:-1]

    response_message = messages[-1]

    # --------------------------------------------------------
    # Tokenize prompt using chat template
    # with assistant generation prompt.
    # --------------------------------------------------------

    prompt_text = (
        tokenizer.apply_chat_template(
            prompt_messages,
            tokenize=False,
            add_generation_prompt=True,
        )
    )

    prompt_ids = tokenizer(
        prompt_text,
        add_special_tokens=False,
        truncation=False,
    )["input_ids"]

    # --------------------------------------------------------
    # Tokenize complete conversation.
    #
    # Because the prompt prefix is identical,
    # full_ids[:len(prompt_ids)] represents
    # the non-trainable prompt portion.
    # --------------------------------------------------------

    full_text = (
        tokenizer.apply_chat_template(
            messages,
            tokenize=False,
            add_generation_prompt=False,
        )
    )

    full_ids = tokenizer(
        full_text,
        add_special_tokens=False,
        truncation=False,
    )["input_ids"]

    # --------------------------------------------------------
    # Extract ONLY assistant response tokens.
    # --------------------------------------------------------

    prompt_len = min(
        len(prompt_ids),
        len(full_ids),
    )

    response_ids = full_ids[
        prompt_len:
    ]

    # --------------------------------------------------------
    # Safety:
    # Never create an example without
    # trainable response tokens.
    # --------------------------------------------------------

    if len(response_ids) == 0:

        raise ValueError(
            "Example contains no assistant "
            "response tokens."
        )

    # --------------------------------------------------------
    # If response itself is larger than MAX_LENGTH,
    # preserve the beginning of the response.
    # --------------------------------------------------------

    if len(response_ids) >= MAX_LENGTH:

        response_ids = response_ids[
            :MAX_LENGTH
        ]

        input_ids = response_ids

        labels = response_ids.copy()

    else:

        # ----------------------------------------------------
        # Keep the RESPONSE completely.
        #
        # If prompt is too long, keep only the
        # LAST prompt tokens.
        # ----------------------------------------------------

        available_prompt = (
            MAX_LENGTH - len(response_ids)
        )

        prompt_ids_trimmed = (
            prompt_ids[-available_prompt:]
        )

        input_ids = (
            prompt_ids_trimmed
            + response_ids
        )

        labels = (
            [-100] * len(
                prompt_ids_trimmed
            )
            + response_ids.copy()
        )

    # --------------------------------------------------------
    # Attention mask
    # --------------------------------------------------------

    attention_mask = [
        1
        for _ in input_ids
    ]

    # --------------------------------------------------------
    # Final safety check
    # --------------------------------------------------------

    if len(input_ids) == 0:

        raise ValueError(
            "Empty tokenized example."
        )

    if not any(
        x != -100
        for x in labels
    ):

        raise ValueError(
            "No trainable response tokens."
        )

    # --------------------------------------------------------
    # Make absolutely sure MAX_LENGTH is respected.
    # --------------------------------------------------------

    input_ids = input_ids[
        :MAX_LENGTH
    ]

    attention_mask = attention_mask[
        :MAX_LENGTH
    ]

    labels = labels[
        :MAX_LENGTH
    ]

    return {
        "input_ids": input_ids,
        "attention_mask": attention_mask,
        "labels": labels,
    }


# ============================================================
# DRIVE CHECKPOINT CALLBACK
# ============================================================

class DriveCheckpointCallback(
    TrainerCallback
):

    def on_save(
        self,
        args,
        state,
        control,
        **kwargs,
    ):

        checkpoint_path = os.path.join(
            args.output_dir,
            f"checkpoint-{state.global_step}",
        )

        distributed = (
            torch.distributed.is_available()
            and torch.distributed.is_initialized()
        )

        if distributed:

            rank = (
                torch.distributed.get_rank()
            )

            if rank == 0:

                if os.path.exists(
                    checkpoint_path
                ):

                    upload_checkpoint_to_drive(
                        checkpoint_path,
                        state.global_step,
                    )

            torch.distributed.barrier()

        else:

            if os.path.exists(
                checkpoint_path
            ):

                upload_checkpoint_to_drive(
                    checkpoint_path,
                    state.global_step,
                )

        return control


# ============================================================
# MAIN
# ============================================================

def main():

    local_rank = int(
        os.environ.get(
            "LOCAL_RANK",
            "0",
        )
    )

    torch.cuda.set_device(
        local_rank
    )

    print("=" * 70)
    print(
        "LLM JUDGE — QWEN 7B STAGE 1"
    )
    print("=" * 70)

    print(
        "LOCAL_RANK:",
        local_rank,
    )

    print(
        "GPU:",
        torch.cuda.get_device_name(
            local_rank
        ),
    )

    print(
        "GPU COUNT:",
        torch.cuda.device_count(),
    )

    # --------------------------------------------------------
    # TOKENIZER
    # --------------------------------------------------------

    print()
    print("=" * 70)
    print("LOADING TOKENIZER")
    print("=" * 70)

    tokenizer = AutoTokenizer.from_pretrained(
        MODEL_PATH,
        local_files_only=True,
        trust_remote_code=True,
        use_fast=True,
    )

    if tokenizer.pad_token is None:

        tokenizer.pad_token = (
            tokenizer.eos_token
        )

    tokenizer.padding_side = "right"

    # --------------------------------------------------------
    # DATASET
    # --------------------------------------------------------

    print()
    print("=" * 70)
    print("LOADING DATASET")
    print("=" * 70)

    dataset = load_from_disk(
        TRAIN_PATH
    )

    print(dataset)

    # --------------------------------------------------------
    # TOKENIZATION
    # --------------------------------------------------------

    print()
    print("=" * 70)
    print("TOKENIZING DATASET")
    print("=" * 70)

    tokenized_dataset = dataset.map(
        lambda x: tokenize_example(
            x,
            tokenizer,
        ),
        remove_columns=dataset.column_names,
        desc="Tokenizing",
    )

    print(
        "Tokenization complete."
    )

    # --------------------------------------------------------
    # MODEL
    # --------------------------------------------------------

    print()
    print("=" * 70)
    print("LOADING 4-BIT MODEL")
    print("=" * 70)

    compute_dtype = torch.float16

    bnb_config = BitsAndBytesConfig(
        load_in_4bit=True,
        bnb_4bit_quant_type="nf4",
        bnb_4bit_compute_dtype=(
            compute_dtype
        ),
        bnb_4bit_use_double_quant=True,
    )

    model = AutoModelForCausalLM.from_pretrained(
        MODEL_PATH,
        quantization_config=bnb_config,
        device_map={
            "": local_rank
        },
        dtype=compute_dtype,
        local_files_only=True,
        trust_remote_code=True,
    )

    model.config.use_cache = False

    # --------------------------------------------------------
    # LORA
    # --------------------------------------------------------

    print()
    print("=" * 70)
    print("ADDING LORA")
    print("=" * 70)

    lora_config = LoraConfig(
        r=8,
        lora_alpha=16,
        lora_dropout=0.05,
        bias="none",
        task_type="CAUSAL_LM",
        target_modules=[
            "q_proj",
            "k_proj",
            "v_proj",
            "o_proj",
        ],
    )

    model = get_peft_model(
        model,
        lora_config,
    )

    model.print_trainable_parameters()

    model.enable_input_require_grads()

    # --------------------------------------------------------
    # TRAINING ARGUMENTS
    # --------------------------------------------------------

    print()
    print("=" * 70)
    print("TRAINING CONFIGURATION")
    print("=" * 70)

    training_args = TrainingArguments(
        output_dir=OUTPUT_DIR,

        num_train_epochs=NUM_EPOCHS,

        per_device_train_batch_size=BATCH_SIZE,

        gradient_accumulation_steps=(
            GRADIENT_ACCUMULATION_STEPS
        ),

        learning_rate=LEARNING_RATE,

        warmup_steps=WARMUP_STEPS,

        logging_steps=10,

        save_strategy="steps",

        save_steps=SAVE_STEPS,

        save_total_limit=None,

        fp16=True,

        bf16=False,

        gradient_checkpointing=True,

        optim="paged_adamw_8bit",

        ddp_find_unused_parameters=False,

        report_to="none",

        remove_unused_columns=False,

        dataloader_num_workers=2,

        save_safetensors=True,
    )

    # --------------------------------------------------------
    # COLLATOR
    # --------------------------------------------------------

    data_collator = JudgeDataCollator(
        tokenizer
    )

    # --------------------------------------------------------
    # TRAINER
    # --------------------------------------------------------

    trainer = Trainer(
        model=model,
        args=training_args,
        train_dataset=tokenized_dataset,
        data_collator=data_collator,
        callbacks=[
            DriveCheckpointCallback()
        ],
    )

    # --------------------------------------------------------
    # RESUME CHECKPOINT
    # --------------------------------------------------------

    resume_checkpoint = None

    if local_rank == 0:

        print()
        print("=" * 70)
        print(
            "CHECKING DRIVE FOR CHECKPOINT"
        )
        print("=" * 70)

        resume_checkpoint = (
            download_latest_checkpoint()
        )

    # --------------------------------------------------------
    # BROADCAST CHECKPOINT
    # --------------------------------------------------------

    if (
        torch.distributed.is_available()
        and torch.distributed.is_initialized()
    ):

        object_list = [
            resume_checkpoint
        ]

        torch.distributed.broadcast_object_list(
            object_list,
            src=0,
        )

        resume_checkpoint = (
            object_list[0]
        )

    # --------------------------------------------------------
    # TRAIN
    # --------------------------------------------------------

    print()
    print("=" * 70)

    if resume_checkpoint:

        print(
            "RESUMING FROM:",
            resume_checkpoint,
        )

    else:

        print(
            "STARTING TRAINING FROM SCRATCH"
        )

    print("=" * 70)

    trainer.train(
        resume_from_checkpoint=(
            resume_checkpoint
            if resume_checkpoint
            else None
        )
    )

    # --------------------------------------------------------
    # FINAL SAVE
    # --------------------------------------------------------

    print()
    print("=" * 70)
    print("SAVING FINAL MODEL")
    print("=" * 70)

    trainer.save_model(
        OUTPUT_DIR
    )

    tokenizer.save_pretrained(
        OUTPUT_DIR
    )

    # --------------------------------------------------------
    # FINAL MODEL → DRIVE
    # --------------------------------------------------------

    if local_rank == 0:

        final_dir = os.path.join(
            OUTPUT_DIR,
            "final",
        )

        if os.path.exists(
            final_dir
        ):

            shutil.rmtree(
                final_dir
            )

        os.makedirs(
            final_dir,
            exist_ok=True,
        )

        for filename in os.listdir(
            OUTPUT_DIR
        ):

            source = os.path.join(
                OUTPUT_DIR,
                filename,
            )

            if os.path.isfile(source):

                shutil.copy2(
                    source,
                    final_dir,
                )

        drive = get_drive_service()

        print()
        print("=" * 70)
        print(
            "UPLOADING FINAL MODEL TO DRIVE"
        )
        print("=" * 70)

        final_folder_id = (
            create_drive_folder(
                drive,
                "final",
                STAGE1_FOLDER_ID,
            )
        )

        for root, _, files in os.walk(
            final_dir
        ):

            for filename in files:

                local_file = os.path.join(
                    root,
                    filename,
                )

                upload_file(
                    drive,
                    local_file,
                    final_folder_id,
                    filename,
                )

        print()
        print(
            "✅ FINAL MODEL UPLOADED TO DRIVE"
        )

    print()
    print("=" * 70)
    print("TRAINING COMPLETE")
    print("=" * 70)


if __name__ == "__main__":
    main()
```

```python
MAX_LENGTH = 768

def tokenize_example(example, tokenizer):
    messages = example["messages"]

    # Dataset structure: system -> user -> assistant
    prompt_messages = messages[:-1]
    assistant_message = messages[-1]

    # Tokenize the prompt and add the assistant generation marker.
    prompt_ids = tokenizer.apply_chat_template(
        prompt_messages,
        tokenize=True,
        add_generation_prompt=True,
        truncation=False,
    )

    # Tokenize ONLY the assistant response.
    response_ids = tokenizer(
        assistant_message["content"],
        add_special_tokens=False,
        truncation=False,
    )["input_ids"]

    if len(response_ids) == 0:
        raise ValueError("Assistant response tokenized to zero tokens.")

    # Keep the complete assistant response whenever possible.
    if len(response_ids) >= MAX_LENGTH:
        input_ids = response_ids[:MAX_LENGTH]
        labels = response_ids[:MAX_LENGTH]

    else:
        prompt_keep = MAX_LENGTH - len(response_ids)

        # Keep the most recent part of the prompt.
        prompt_part = prompt_ids[-prompt_keep:]

        input_ids = prompt_part + response_ids

        # Train only on the assistant response.
        labels = [-100] * len(prompt_part) + response_ids

    attention_mask = [1] * len(input_ids)

    # Safety check.
    if not any(label != -100 for label in labels):
        raise ValueError("Example contains no trainable assistant tokens.")

    return {
        "input_ids": input_ids,
        "attention_mask": attention_mask,
        "labels": labels,
    }
```

```python
import json
import re
import time
import os
import pandas as pd

TECH_PATH = "/kaggle/working/LLM-Judge/data/evaluation/technical_50.json"
RESULT_PATH = "/kaggle/working/LLM-Judge/data/evaluation/technical_50_finetuned_results.json"


def parse_judge_response(response):
    """Extract {"A": "...", "B": "..."} from model output."""
    
    match = re.search(r'\{.*?\}', response, re.DOTALL)
    
    if not match:
        return None
    
    try:
        result = json.loads(match.group())
    except json.JSONDecodeError:
        return None
    
    if "A" not in result or "B" not in result:
        return None
    
    a = str(result["A"]).lower().strip()
    b = str(result["B"]).lower().strip()
    
    if a not in {"correct", "incorrect"}:
        return None
    
    if b not in {"correct", "incorrect"}:
        return None
    
    return {
        "A": a,
        "B": b
    }


def get_winner(parsed):
    """Convert correctness labels into A/B/TIE."""
    
    if parsed is None:
        return None
    
    if parsed["A"] == "correct" and parsed["B"] == "incorrect":
        return "A"
    
    if parsed["A"] == "incorrect" and parsed["B"] == "correct":
        return "B"
    
    return "TIE"


def evaluate_once(problem, answer_a, answer_b):
    """Run one model evaluation."""
    
    messages = [
        {
            "role": "system",
            "content": """You are an impartial AI response judge.
Evaluate both answers independently.

For each answer, classify it as exactly:
- "correct"
- "incorrect"

Return ONLY valid JSON in this exact format:
{"A":"correct","B":"incorrect"}"""
        },
        {
            "role": "user",
            "content": f"""PROBLEM:
{problem}

ANSWER A:
{answer_a}

ANSWER B:
{answer_b}

Evaluate both answers."""
        }
    ]
    
    prompt = tokenizer.apply_chat_template(
        messages,
        tokenize=False,
        add_generation_prompt=True
    )
    
    inputs = tokenizer(
        prompt,
        return_tensors="pt",
        truncation=True,
        max_length=4096
    ).to(model.device)
    
    start = time.perf_counter()
    
    with torch.no_grad():
        outputs = model.generate(
            **inputs,
            max_new_tokens=50,
            do_sample=False,
            pad_token_id=tokenizer.eos_token_id,
        )
    
    latency = time.perf_counter() - start
    
    generated = outputs[0][inputs["input_ids"].shape[1]:]
    
    response = tokenizer.decode(
        generated,
        skip_special_tokens=True
    ).strip()
    
    parsed = parse_judge_response(response)
    
    return {
        "raw_response": response,
        "parsed": parsed,
        "winner": get_winner(parsed),
        "latency_seconds": round(latency, 3),
        "valid": parsed is not None
    }


# --------------------------------------------------
# Load dataset
# --------------------------------------------------

with open(TECH_PATH, "r", encoding="utf-8") as f:
    dataset = json.load(f)

print("=" * 70)
print("POST-FINE-TUNING TECHNICAL BENCHMARK")
print("=" * 70)
print(f"Dataset size: {len(dataset)}")
print("Evaluations: 100")
print("Model: Qwen2.5-7B-Instruct + LoRA")
print("=" * 70)


results = []

for index, item in enumerate(dataset, start=1):
    
    print(
        f"[{index:02d}/{len(dataset)}] "
        f"{item['id']} | "
        f"{item['category']} | "
        f"{item['difficulty']}",
        flush=True
    )
    
    try:
        # -----------------------------
        # Original position
        # Good = A, Bad = B
        # Expected winner = A
        # -----------------------------
        
        original = evaluate_once(
            problem=item["problem"],
            answer_a=item["good_answer"],
            answer_b=item["bad_answer"]
        )
        
        # -----------------------------
        # Reversed position
        # Bad = A, Good = B
        # Expected winner = B
        # -----------------------------
        
        reversed_result = evaluate_once(
            problem=item["problem"],
            answer_a=item["bad_answer"],
            answer_b=item["good_answer"]
        )
        
        results.append({
            "id": item["id"],
            "category": item["category"],
            "difficulty": item["difficulty"],
            "original": {
                **original,
                "expected": "A",
                "correct": original["winner"] == "A"
            },
            "reversed": {
                **reversed_result,
                "expected": "B",
                "correct": reversed_result["winner"] == "B"
            }
        })
        
        print(
            f"   Original: {original['winner']} "
            f"({'✓' if original['winner'] == 'A' else '✗'}) | "
            f"Reversed: {reversed_result['winner']} "
            f"({'✓' if reversed_result['winner'] == 'B' else '✗'})"
        )

    except Exception as e:
        print(f"   ERROR: {e}")
        
        results.append({
            "id": item["id"],
            "category": item["category"],
            "difficulty": item["difficulty"],
            "error": str(e)
        })


# --------------------------------------------------
# Calculate metrics
# --------------------------------------------------

successful = [
    r for r in results
    if "error" not in r
]

total_pairs = len(results)
successful_pairs = len(successful)
failed_pairs = total_pairs - successful_pairs

original_correct = sum(
    r["original"]["correct"]
    for r in successful
)

reversed_correct = sum(
    r["reversed"]["correct"]
    for r in successful
)

original_valid = sum(
    r["original"]["valid"]
    for r in successful
)

reversed_valid = sum(
    r["reversed"]["valid"]
    for r in successful
)

position_consistent = sum(
    r["original"]["winner"] == "A"
    and r["reversed"]["winner"] == "B"
    for r in successful
)

first_position_bias = sum(
    r["original"]["winner"] == "A"
    and r["reversed"]["winner"] == "A"
    for r in successful
)

second_position_bias = sum(
    r["original"]["winner"] == "B"
    and r["reversed"]["winner"] == "B"
    for r in successful
)

latencies = [
    r["original"]["latency_seconds"]
    for r in successful
] + [
    r["reversed"]["latency_seconds"]
    for r in successful
]

metrics = {
    "total_pairs": total_pairs,
    "successful_pairs": successful_pairs,
    "failed_pairs": failed_pairs,
    "total_evaluations": successful_pairs * 2,
    
    "original_accuracy": round(
        original_correct / successful_pairs, 4
    ) if successful_pairs else 0,
    
    "reversed_accuracy": round(
        reversed_correct / successful_pairs, 4
    ) if successful_pairs else 0,
    
    "position_consistency": round(
        position_consistent / successful_pairs, 4
    ) if successful_pairs else 0,
    
    "first_position_bias": first_position_bias,
    "second_position_bias": second_position_bias,
    
    "position_bias_rate": round(
        (first_position_bias + second_position_bias)
        / successful_pairs,
        4
    ) if successful_pairs else 0,
    
    "original_valid_rate": round(
        original_valid / successful_pairs, 4
    ) if successful_pairs else 0,
    
    "reversed_valid_rate": round(
        reversed_valid / successful_pairs, 4
    ) if successful_pairs else 0,
    
    "average_latency_seconds": round(
        sum(latencies) / len(latencies),
        3
    ) if latencies else 0
}


# --------------------------------------------------
# Save results
# --------------------------------------------------

output = {
    "benchmark": "technical_50",
    "model": "Qwen2.5-7B-Instruct + LoRA",
    "metrics": metrics,
    "results": results
}

with open(RESULT_PATH, "w", encoding="utf-8") as f:
    json.dump(output, f, indent=2, ensure_ascii=False)


# --------------------------------------------------
# Display final metrics
# --------------------------------------------------

print("\n")
print("=" * 70)
print("TECHNICAL 50 — POST-FINE-TUNING RESULTS")
print("=" * 70)

for key, value in metrics.items():
    print(f"{key:30s}: {value}")

print("=" * 70)
print(f"Results saved to:\n{RESULT_PATH}")
print("=" * 70)
```

```python
import json
import re
import time
import os
import torch

HARD_PATH = "/kaggle/working/LLM-Judge/data/evaluation/hard_technical_100.json"
RESULT_PATH = "/kaggle/working/LLM-Judge/data/evaluation/hard_technical_100_finetuned_results.json"


def parse_judge_response(response):
    match = re.search(r'\{.*?\}', response, re.DOTALL)

    if not match:
        return None

    try:
        result = json.loads(match.group())
    except json.JSONDecodeError:
        return None

    if "A" not in result or "B" not in result:
        return None

    a = str(result["A"]).lower().strip()
    b = str(result["B"]).lower().strip()

    if a not in {"correct", "incorrect"}:
        return None

    if b not in {"correct", "incorrect"}:
        return None

    return {"A": a, "B": b}


def get_winner(parsed):
    if parsed is None:
        return None

    if parsed["A"] == "correct" and parsed["B"] == "incorrect":
        return "A"

    if parsed["A"] == "incorrect" and parsed["B"] == "correct":
        return "B"

    return "TIE"


def evaluate_once(problem, answer_a, answer_b):

    messages = [
        {
            "role": "system",
            "content": """You are an impartial AI response judge.
Evaluate both answers independently.

For each answer, classify it as exactly:
- "correct"
- "incorrect"

Return ONLY valid JSON in this exact format:
{"A":"correct","B":"incorrect"}"""
        },
        {
            "role": "user",
            "content": f"""PROBLEM:
{problem}

ANSWER A:
{answer_a}

ANSWER B:
{answer_b}

Evaluate both answers."""
        }
    ]

    prompt = tokenizer.apply_chat_template(
        messages,
        tokenize=False,
        add_generation_prompt=True
    )

    inputs = tokenizer(
        prompt,
        return_tensors="pt",
        truncation=True,
        max_length=4096
    ).to(model.device)

    start = time.perf_counter()

    with torch.no_grad():
        outputs = model.generate(
            **inputs,
            max_new_tokens=50,
            do_sample=False,
            pad_token_id=tokenizer.eos_token_id,
        )

    latency = time.perf_counter() - start

    generated = outputs[0][inputs["input_ids"].shape[1]:]

    response = tokenizer.decode(
        generated,
        skip_special_tokens=True
    ).strip()

    parsed = parse_judge_response(response)

    return {
        "raw_response": response,
        "parsed": parsed,
        "winner": get_winner(parsed),
        "latency_seconds": round(latency, 3),
        "valid": parsed is not None
    }


# --------------------------------------------------
# Load Hard 100
# --------------------------------------------------

with open(HARD_PATH, "r", encoding="utf-8") as f:
    dataset = json.load(f)

print("=" * 70)
print("POST-FINE-TUNING HARD 100 BENCHMARK")
print("=" * 70)
print(f"Dataset size: {len(dataset)}")
print("Evaluations: 200")
print("Model: Qwen2.5-7B-Instruct + LoRA")
print("=" * 70)


results = []

for index, item in enumerate(dataset, start=1):

    print(
        f"[{index:03d}/{len(dataset)}] "
        f"{item['id']} | "
        f"{item['category']} | "
        f"{item['difficulty']}",
        flush=True
    )

    try:

        # Original:
        # Good = A, Bad = B
        original = evaluate_once(
            problem=item["problem"],
            answer_a=item["good_answer"],
            answer_b=item["bad_answer"]
        )

        # Reversed:
        # Bad = A, Good = B
        reversed_result = evaluate_once(
            problem=item["problem"],
            answer_a=item["bad_answer"],
            answer_b=item["good_answer"]
        )

        results.append({
            "id": item["id"],
            "category": item["category"],
            "difficulty": item["difficulty"],

            "original": {
                **original,
                "expected": "A",
                "correct": original["winner"] == "A"
            },

            "reversed": {
                **reversed_result,
                "expected": "B",
                "correct": reversed_result["winner"] == "B"
            }
        })

        print(
            f"   Original: {original['winner']} "
            f"({'✓' if original['winner'] == 'A' else '✗'}) | "
            f"Reversed: {reversed_result['winner']} "
            f"({'✓' if reversed_result['winner'] == 'B' else '✗'})"
        )

    except Exception as e:

        print(f"   ERROR: {e}")

        results.append({
            "id": item["id"],
            "category": item["category"],
            "difficulty": item["difficulty"],
            "error": str(e)
        })


# --------------------------------------------------
# Metrics
# --------------------------------------------------

successful = [
    r for r in results
    if "error" not in r
]

successful_pairs = len(successful)

original_correct = sum(
    r["original"]["correct"]
    for r in successful
)

reversed_correct = sum(
    r["reversed"]["correct"]
    for r in successful
)

original_valid = sum(
    r["original"]["valid"]
    for r in successful
)

reversed_valid = sum(
    r["reversed"]["valid"]
    for r in successful
)

position_consistent = sum(
    r["original"]["winner"] == "A"
    and r["reversed"]["winner"] == "B"
    for r in successful
)

first_position_bias = sum(
    r["original"]["winner"] == "A"
    and r["reversed"]["winner"] == "A"
    for r in successful
)

second_position_bias = sum(
    r["original"]["winner"] == "B"
    and r["reversed"]["winner"] == "B"
    for r in successful
)

latencies = [
    r["original"]["latency_seconds"]
    for r in successful
] + [
    r["reversed"]["latency_seconds"]
    for r in successful
]


metrics = {
    "total_pairs": len(results),
    "successful_pairs": successful_pairs,
    "failed_pairs": len(results) - successful_pairs,
    "total_evaluations": successful_pairs * 2,

    "original_accuracy": round(
        original_correct / successful_pairs, 4
    ) if successful_pairs else 0,

    "reversed_accuracy": round(
        reversed_correct / successful_pairs, 4
    ) if successful_pairs else 0,

    "position_consistency": round(
        position_consistent / successful_pairs, 4
    ) if successful_pairs else 0,

    "first_position_bias": first_position_bias,
    "second_position_bias": second_position_bias,

    "position_bias_rate": round(
        (first_position_bias + second_position_bias)
        / successful_pairs,
        4
    ) if successful_pairs else 0,

    "original_valid_rate": round(
        original_valid / successful_pairs, 4
    ) if successful_pairs else 0,

    "reversed_valid_rate": round(
        reversed_valid / successful_pairs, 4
    ) if successful_pairs else 0,

    "average_latency_seconds": round(
        sum(latencies) / len(latencies),
        3
    ) if latencies else 0
}


# --------------------------------------------------
# Save
# --------------------------------------------------

output = {
    "benchmark": "hard_technical_100",
    "model": "Qwen2.5-7B-Instruct + LoRA",
    "metrics": metrics,
    "results": results
}

with open(RESULT_PATH, "w", encoding="utf-8") as f:
    json.dump(output, f, indent=2, ensure_ascii=False)


# --------------------------------------------------
# Final output
# --------------------------------------------------

print("\n")
print("=" * 70)
print("HARD 100 — POST-FINE-TUNING RESULTS")
print("=" * 70)

for key, value in metrics.items():
    print(f"{key:30s}: {value}")

print("=" * 70)
print(f"Results saved to:")
print(RESULT_PATH)
print("=" * 70)
```
